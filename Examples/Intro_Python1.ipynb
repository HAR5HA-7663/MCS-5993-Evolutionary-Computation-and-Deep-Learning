## iterator
* an object used to iterate over iterable objects
* commonly used to loop through the objects of a collection, read them, and *remove* them

```
fruits = ["apple", "banana", "cherry", "berry", "mellon"]
iterator1 = (f.upper() for f in fruits) # note ( ... )
print(next(iterator1)) # the first item processed (printed) and removed
for f in iterator1: # the rest items are processed
  print(f, end=' ')
#print(next(iterator1)) #StopIteration Exception will occur
```

```
# another way to create an iterator using iter()
iterator2 = iter(fruits)
print(next(iterator2).upper())
print(next(iterator2).upper())
for f in iterator2: # the rest items are processed
  print(f.upper(), end=' ')
```

## Looping with "while" and **formatting**

```
n = 10

sum = 0
counter = 1
while counter <= n:
    sum = sum + counter
    counter += 1

print("Sum of 1 + ... + %d = %d" % (n,sum)) # old style formatting
print("Sum of 1 + ... + {0} = {1}".format(n, sum))
print(f"Sum of 1 + ... + {n} = {sum}")  # Python 3.6+
```

```
a = 3.141592
print(f"{a:.3f}")
print(f"{a:5.2f}") # width.precision.
                   # width: the # of characters used in total to display
```

## Exception Handling
[Python Exception names](https://www.tutorialspoint.com/python/python_exceptions.htm)

```
while True:
  try:
    n = int(input("Enter a number: "))
    for i in range(n):
      for j in range (i+1):
        print('X', end='')
      print()
    break     # break the while loop
  #except ValueError: Raised when an operation or function receives an argument that has the right type but an inappropriate value
  except Exception:
    print("That's not a number. Try again") # continue
```

```
while True:
  try:
    n = int(input("Enter a positive number: "))
    if (n < 1):
      raise Exception(f'{n} is not a positive int.')
    for i in range(n):
      for j in range (i+1):
        print('X', end='')
      print()
    break     # break the while loop
  except Exception as e:
    print(f"{e}") # continue
```
