## Local & Global Variables
Note: The problem with global variables is that since every function has access to these, it becomes increasingly hard to figure out which functions actually read and write these variables.

In ROS design, gobal variables are usually used for state vaiables when nodes are supposed to be small.

```
a = 1

def f():
    print('Inside f(): ', a)  # Uses global because there is no local 'a'
    print()

def g():
    a = 2
    print(f'Inside g(): {a}') # Variable 'a' is redefined as a local
    print()

def h():
    global a
    a = 3
    print(f'Inside h(): {a}') # Uses global keyword to modify global 'a'
    print()

# Global scope
print('global : ', a)
f()
print('global : ', a)
g()
print('global : ', a)
h()

print('global : ', a)
```

```
# Note that x is not defined outside functions
def f1():
    global x
    x = 3
    print(f'Inside f1(): {x}')

def f2():
    x = 5
    print(f'Inside f2(): {x}')

f1()
f2()
print(f'{x}')
```

```
def f1():
    global x
    x = 3

def f2():
    x = 0
    print(f'{x}')

f1()
f2()
```

```
name = 'cj'
def example():
  print(name)

example()
```

```
# what's wrong with this code?
name = 'Rob'
def example():
  print(name)                       # <=== unbounded error
  name = 'Bob'

example()
```

```
# fixed
name = 'Rob'
def example():
  global name
  print(name)
  name = 'Bob'

example()
print(name)
```

```
# 2 prints
n = 3
def func():
  global n
  print(n)
  n = 8
  print(n)

func()
```

```
# 2 funcs
def func1():
  global n
  n = 8

def func2():
  global n
  n = n + 2
  print(n)

func1()
func2()
print(n)
```

## Inner (Nested) Functions

## Classes, Objects, and Methods

```
# Create a Math class
class MyMath:
  def plus(self, a, b):
    self.ans = a + b
    return self.ans
  def multiply(self, a, b):
    self.ans = a * b
    return self.ans

o = MyMath() # instantiate an object that has + and * methods
print(o.plus(2,3))
print(o.multiply(2,3))
```

```
class Person:
  def __init__(self, name, salary):
    self.name = name
    self.salary = salary

  def myGetMethod(self):
    print(f"Hello my name is {self.name}")

  def mySetMethod(self, n):
    self.salary += n
    print(f"My adjusted salary is ${self.salary}")

p1 = Person("John", 8000)
p1.myGetMethod()
p1.mySetMethod(300)
```
