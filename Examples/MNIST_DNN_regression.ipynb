```python
from keras import layers, models
from keras.datasets import mnist
from sklearn.model_selection import train_test_split
```

```python
# Load and preprocess the MNIST dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train, x_val, y_train, y_val = train_test_split(x_train, y_train, test_size=0.1, random_state=42)

# Normalize pixel values to be between 0 and 1
x_train, x_val, x_test = x_train / 255.0, x_val / 255.0, x_test / 255.0

# Flatten the images. Q: Do we need to flatten???

x_train = x_train.reshape((x_train.shape[0], -1))
x_val = x_val.reshape((x_val.shape[0], -1))
x_test2 = x_test.reshape((x_test.shape[0], -1)) # x_test is saved for display
```

```python
# Define the model
model = models.Sequential([
    layers.Dense(512, activation='relu', input_shape=(784,)), # DNN needs flattened
    layers.Dense(128, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)  # One output neuron for regression
])
```

```python
# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])

# Train the model
model.fit(x_train, y_train, epochs=20, batch_size=32, validation_data=(x_val, y_val))

```

```python
# Evaluate the model on the test set
test_loss, test_mae = model.evaluate(x_test2, y_test)
print(f'Test Mean Squared Error: {test_loss}, Test Mean Absolute Error: {test_mae}')
```
