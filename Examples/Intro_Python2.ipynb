## Passing Arguments
Very often the terms parameter and argument are used synonymously, but there is a clear difference. Parameters are inside functions, while arguments are used in function calls, i.e. the values passed to the function at run-time.

*   If you pass immutable arguments like integers, strings or tuples to a function, the passing acts like call-by-value. The object reference is passed to the function parameters. They can't be changed within the function, because they can't be changed at all, i.e. they are immutable.

*   Mutable arguments are also passed by object reference, but they can be changed in place within the function.


https://python-course.eu/python-tutorial/passing-arguments.php

```
def sumAfterAppend(list, num):
  list.append(num) #list3 modified
  sum = 0
  for x in list:
    sum += x
  return sum

list3 = [1,2,3]
print (sumAfterAppend(list3, 100))
print(list3)
```

```
def add_one(x):   # numerics being immutable
    x += 1        # new object created
    return x
  
def append_one(x): # lists being mutable
    x += [1]
    return x

a = 23
print(id(a) == id(add_one(a)))
    
a = [1,2,3]
print(id(a) == id(append_one(a)))
print(a)
    
```

## Join()
The join() method provides a flexible way to concatenate string. It concatenates each element of an iterable (such as list, string and tuple) to the string and returns the concatenated string. The syntax of join is:

string.join(iterable)


```
numList = ['1', '2', '3', '4']
seperator = ', '
print(seperator.join(numList))

numTuple = ('a', 'b', 'c', 'd')
print(seperator.join(numTuple))

```

```
s1 = 'abc'
s2 = '123'

""" Each character of s2 is concatenated to the front of s1""" 
print('s1.join(s2):', s1.join(s2))

""" Each character of s1 is concatenated to the front of s2""" 
print('s2.join(s1):', s2.join(s1))
```

#List comprehensions
provide a concise way to **create** lists based on existing iterables 

```
numbers = [1,2,3,4,5,6,7]
squared = [x*x for x in numbers]
print(squared)
```

```
# Using for loop
squared =[]
for i in numbers:
  squared.append(i*i)
print(squared)
```

```
numbers = [1,2,3,4,5,6,7]
evens = [x for x in numbers if x % 2 == 0]
print(evens)
```

```
evens = []
for i in numbers:
   if i % 2 == 0:
      evens.append(i)
print(evens)
```

```
odds = [y for y in numbers if y not in evens]
print(odds)
```

#Lambda Function
a function that is defined without a name. We use lambda functions when we require a nameless function for a short period of time.

```
sq = lambda x: x*x
print(sq(6))
```

```
# to filter out only even numbers from a list, using a lambda f 
my_list = [1,2,3,4,5,6,7,8,9]
even_list = list(filter(lambda x: (x%2 == 0), my_list))
print(even_list)
```
