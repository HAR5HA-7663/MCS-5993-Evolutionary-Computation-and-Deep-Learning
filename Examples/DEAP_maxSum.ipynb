```python
import random
from deap import base, creator, tools, algorithms

# Step 1: Define the evaluation function
def eval_individual(individual):
    return sum(individual),  # The comma makes it a tuple, which DEAP expects

# Step 2: Set up DEAP components
# Create a fitness function to maximize
creator.create("FitnessMax", base.Fitness, weights=(1.0,))
# Create the individual, which is a list of integers
creator.create("Individual", list, fitness=creator.FitnessMax)

# Initialize the toolbox
toolbox = base.Toolbox()

# Register the attributes for each gene (random integers between 0 and 100, inclusive)
toolbox.register("attr_int", random.randint, 0, 100)
# Create an individual with 10 attributes
toolbox.register("individual", tools.initRepeat, creator.Individual, toolbox.attr_int, 10)
# Create a population of 50 individuals
toolbox.register("population", tools.initRepeat, list, toolbox.individual, 10)

# Register the evaluation, crossover, mutation, and selection functions
toolbox.register("evaluate", eval_individual)
toolbox.register("mate", tools.cxTwoPoint)
toolbox.register("mutate", tools.mutFlipBit, indpb=0.05)
toolbox.register("select", tools.selTournament, tournsize=3)

# Step 3: Set up the main function to run the algorithm
def main():
    # Initialize population
    population = toolbox.population()

    # Run the algorithm
    final_population = algorithms.eaSimple(
        population,
        toolbox,
        cxpb=0.5,  # Crossover probability
        mutpb=0.2,  # Mutation probability
        ngen=40,    # Number of generations
        verbose=False,
    )

    # Gather all individuals in the final population
    print(final_population)

    print("Final population:")
    for individual in final_population[0]:
        print(individual)
    top_individual = tools.selBest(final_population[0], 1)[0]
    print("Best individual:", top_individual)
    print("Fitness:", top_individual.fitness.values[0])

main()
```
